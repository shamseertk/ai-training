```python
from pydantic import BaseModel, Field

class ReviewIn(BaseModel):
    product: str
    rating: int = Field(..., ge=1, le=5, description="The rating given in the review (1-5)")
    text: str

good = ReviewIn(product='Ceramic Coffee Mug', rating=5, text='This mug is great! It keeps my coffee hot for hours and the design is beautiful.')

print(good)

print('rating is a ', type(good.rating).__name__)

#throws an error because the rating is out of range (1-5) and text is missing
try :
    ReviewIn(product='Ceramic Coffee Mug', rating=6)
except Exception as e:
    print(e)
```

```python
from typing import Literal

class PredictRequest(BaseModel):
    text: str = Field(min_length=1, description="empty strings are not allowed")

class PredictResponse(BaseModel):
    label: Literal['positive', 'negative']
    score: float = Field(..., ge=0, le=1, description="The confidence score of the prediction (0-1)")

req = PredictRequest(text='This mug is great! It keeps my coffee hot for hours and the design is beautiful.')
resp = PredictResponse(label='positive', score=0.95)
print("Request:", req)
print("Response:", resp)
```

```python
def buggy_model(text: str) -> dict:
  return {"label": "positive", "score": 1.5}  # score is out of range (0-1)

raw = buggy_model("This mug is great! It keeps my coffee hot for hours and the design is beautiful.")

try:
    resp = PredictResponse(**raw)
except Exception as e:
    print(e)
```
