### Load and Explore

```python
import pandas as pd
# read the data from csv file first 5 rows and loads it into a DataFrame(a table in memory)
reviews = pd.read_csv('product_reviews.csv')

#head() shows the first 5 records 
reviews.head()


reviews.info()  # provides a concise summary of the DataFrame, including the number of non-null entries and data types for each column
```

```python
reviews.describe()  # generates descriptive statistics that summarize the central tendency, dispersion, and shape of a dataset’s distribution, excluding NaN values
```

```python
df = reviews.copy() 
print(" Num of rows", len(df))
df = df.drop_duplicates()
print(" Num of rows after dropping duplicates", len(df))
#. the raing of 9 is invalid on a 1-5 scale. Treat it is missing value
df.loc[df['rating'] > 5, 'rating'] = None

# 3. Fill missing ratingwith the medium rating.
median_rating = df['rating'].median()
df['rating'] = df['rating'].fillna(median_rating)
print('median rating:', median_rating)
print('any ratings still missing?', df['rating'].isna().any())
```

```python
# 4. Drop the rows with missing review text.
df = df.dropna(subset=['review_comment'])
print("Number of rows after dropping rows with missing review text:", len(df))

# 5. Proudct names have inconsitent casing.
#convert all product names to lowercase to ensure consistency.
df['product'] = df['product'].str.title()

summary = df.groupby('product')['rating'].agg(['mean', 'count']).round(2)

summary

```

```python
import numpy as np

# quick list-vs-array reminder: the array does the match on every elevement at once
py_list = [1, 2, 3, 4, 5]
print("py_list:", py_list)
print('py_list * 2:', py_list * 2)
print('array * 2 ->', np.array(py_list) * 2)

df["verified_flag"] = (df['verified'] == True).astype(int)
# pick the two numeric columns and to_numpy() to convert them to numpy arrays. Then we can use the np.corrcoef() function to compute the correlation coefficient between the two arrays.
features = df[['rating', 'verified_flag']].to_numpy(dtype=float)
#shape reports the number of rows and columns in the array. The first element of the shape tuple is the number of rows, and the second element is the number of columns.
print("features shape:", features.shape)
features[:5]  # show the first 5 rows of the features array
```
